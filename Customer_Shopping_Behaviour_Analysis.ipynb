*Generally choose Median over Mean since Mean can get affected by outliers*

We cannot simply use Median directly since ratings are not similar for every product 

```python
# Impute missing values using Median for EACH CATEGORY

df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
# Creating a column age_group to group the customer into: young adults, adults, middle age, seniors
labels = ['Young Adult', 'Adult', 'Middle-aged','Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels) # q=4 means 4 Equal parts
```

*q* means *Quantile Cut* i.e *It divides data into equal-sized groups*

**It divides in such a way that each group has equal number of rows**

```python
# Column for purchase_frequency_days -> textual to numeric
frequency_mapping = {
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months': 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
